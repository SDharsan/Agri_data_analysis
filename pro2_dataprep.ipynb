```python
engine = create_engine('mysql+pymysql://root:1015@localhost:3306/Agriculture')
```

```python
df.to_sql('agri_data', con=engine, if_exists='replace', index=False)
```

```python
sql_connection = pymysql.connect(host='localhost', user='root', password='1015', database='Agriculture')
cur = sql_connection.cursor()
```

```python
Year_wise_Trend_of_Rice_Production_Across_States  = cur.execute("select state_name,year,sum(rice_production) as rice_production from agri_data GROUP BY year, state_name ORDER BY year, state_name desc limit 3;")
```

```python
Top_5_Districts_by_Wheat_Yield_Increase_Over_the_Last_5_Years = cur.execute("SELECT dist_name, year,(wheat_yield - LAG(wheat_yield) OVER (PARTITION BY dist_name ORDER BY year)) AS yield_increase FROM agri_data ORDER BY year DESC LIMIT 5;")
```

```python
#States with the Highest Growth in Oilseed Production (5-Year Growth Rate)
Oilseed_Production_Growth = cur.execute("SELECT state_name, ((MAX(oilseeds_production) - MIN(oilseeds_production)) / MIN(oilseeds_production)) * 100 AS growth_rate FROM agri_data GROUP BY state_name ORDER BY growth_rate DESC LIMIT 5;")
```

```python
# Yearly Production Growth of Cotton in Top 5 Cotton Producing States

cotton_production_growth = cur.execute("select state_name, (sum(cotton_production) - lag(sum(cotton_production)) over (partition by state_name order by year)) as yearly_growth from agri_data group by state_name, year order by yearly_growth desc limit 5;")
```

```python
# Districts with the Highest Groundnut Production in 2017
highest_groundnut_2017 = cur.execute("SELECT dist_name, groundnut_production FROM agri_data WHERE year = 2017 ORDER BY groundnut_production DESC LIMIT 5;")
```

```python
# Annual Average Maize Yield Across All States
annual_avg_maize_yield = cur.execute("SELECT state_name, AVG(maize_yield) AS avg_yield FROM agri_data GROUP BY state_name;")
```

```python
# Total Area Cultivated for Oilseeds in Each State
total_oilseed_area = cur.execute("SELECT state_name, SUM(oilseeds_area) AS total_area FROM agri_data GROUP BY state_name;")
```

```python
# Districts with the Highest Rice Yield
highest_rice_yield = cur.execute("SELECT dist_name, rice_yield FROM agri_data ORDER BY rice_yield DESC LIMIT 5;")

```

```python
# Compare the Production of Wheat and Rice for the Top 5 States Over 10 Years
top_5_states = cur.execute("""SELECT
    year,
    state_name,
    SUM(rice_production) AS total_rice_production,
    SUM(wheat_production) AS total_wheat_production
FROM agri_data
WHERE year >= (SELECT MAX(year) - 9 FROM agri_data)
GROUP BY year, state_name
ORDER BY state_name, year;""")
```
